```python
# 3 cols with nulls- region_category,prefered_offer_types,points_in_wallet
data['region_category'].value_counts()
# we cannot just fill it with the mode here since we cant be sure of the persons region,
# it might be inaccurate,which is not very helpful.
# not assuming sometimes is better than bad assumptions
# assigning all missing rows to the most common category (Town) may bias the distribution ,
# it may create fake patterns for the model to learn
```

```python
data['region_category'] = data['region_category'].fillna('Unknown')
```

```python
# prefered_offer_types is not a numerical col as well hence same fill with unknown
data['preferred_offer_types'] = data['preferred_offer_types'].fillna('Unknown')
```

```python
# numerical column, so use median since it is robust to outliers.
data['points_in_wallet'] = data['points_in_wallet'].fillna(data['points_in_wallet'].median())
# check 
data['preferred_offer_types'].value_counts()
```

```python
# drop useless cols which dont play a role in model training
drop_cols = ['Unnamed: 0', 'security_no', 'referral_id']
data.drop(columns=drop_cols,inplace = True)
```

```python
# next step must be handling data and time columns, as the model might confuse it for a string as its datatype is 'str'
# check joining_date,last_visit_time
data[['joining_date','last_visit_time']]
```

```python
data['joining_date']=pd.to_datetime(data['joining_date'],errors='coerce')
data['joining_date']
```

```python
# getting the most recent date
reference_date = data['joining_date'].max()
reference_date
```

```python
# feature engineering from joining_date
# questions to ask!!
# How long has the customer been with the company?
data['tenure_days'] = (reference_date - data['joining_date']).dt.days
```

```python
data['joining_month'] = data['joining_date'].dt.month
data['joining_year'] = data['joining_date'].dt.year
```

```python
data.drop(columns=['joining_date'], inplace=True)
```

```python
# convert to datetime
data['last_visit_time'] = pd.to_datetime(data['last_visit_time'],format='%H:%M:%S', errors='coerce')
```

```python
# since time itself is not that useful for the model to interpret, go ahead with the hour of the day
data['last_visit_hour'] = data['last_visit_time'].dt.hour
```

```python
data.drop(columns=['last_visit_time'], inplace=True)
```

```python
obj_cols = data.select_dtypes(include='object').columns.tolist()

for col in obj_cols:
    data[col] = data[col].astype(str).str.strip()
data.head()
```

```python
# Feature Engineering from Existing Columns
#  Complaint-based features
data['complaint_status'].value_counts()
```

```python
data['past_complaint'] = data['past_complaint'].astype(str).str.lower().str.strip()
data['complaint_status'] = data['complaint_status'].astype(str).str.lower().str.strip()
```

```python
# find out complaints which are unresolved
resolved = ['solved', 'solved in follow up']

data['complaint_unresolved'] = (
    (data['past_complaint'] == 'yes') &
    (~data['complaint_status'].isin(resolved))
).astype(int)
```
